```python
import os
import h5py
from src.utils.data_utils import read_h5
import matplotlib.pyplot as plt
```

```python
# SPH
path = "/home/atoshev/code/sph-turbulence/gen_dataset/data"
# ML dataset
ds_path = "/home/atoshev/code/sph_les/data/2D_KOLM_4096_200kevery1/test.h5"
ds = h5py.File(ds_path, "r")

for i in range(5):
    path_i = os.path.join(path, f"traj_{str(15 + i)}_step_01000.h5")
    frame_i = read_h5(path_i)  # r, u; (4096, 2), (4096, 2)
    frame_j = ds[f"{str(i).zfill(5)}"]  # position, u; (20001, 4096, 2), (20001, 4096, 2)

    fig, axs = plt.subplots(1, 2, figsize=(8, 4))
    axs[0].scatter(frame_i["r"][:, 0], frame_i["r"][:, 1], c=frame_i["u"][:, 0], s=2)
    axs[1].scatter(
        frame_j["position"][0, :, 0], frame_j["position"][0, :, 1], c=frame_j["u"][0, :, 0], s=2
    )
    for ax in axs:
        ax.set_xlim(0, 2 * 3.14159)
        ax.set_ylim(0, 2 * 3.14159)
        ax.set_aspect("equal")
    plt.show()
```
