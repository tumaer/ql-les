# **Rollout visualizations**

This notebook is used to visualize predicted and target trajectories.

Pick a rollout file and animate it.


```python
import pickle
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import os
from IPython.display import Image

import sys

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))

# Specify the relative path to pkl # "logs/rollouts/path_to_pkl"
rollout_path = "../logs/rollouts/w290fhap/test/rollout_0010.pkl"
```

## TGV

```python
# Construct the absolute path
print(f"Absolute path: {rollout_path}")

# Check if the file exists
if not os.path.exists(rollout_path):
    raise FileNotFoundError(f"No such file or directory: '{rollout_path}'")

# Load the rollout data
rollout = pickle.load(open(rollout_path, "rb"))
print(f"Rollout of shape {rollout['predicted_rollout'].shape} (steps, nodes, xy pos)")

# Create the figure and axes
fig, ax = plt.subplots(1, 2)
ax[0].set_xlim([0, 1.0])
ax[0].set_ylim([0, 1.0])
ax[1].set_xlim([0, 1.0])
ax[1].set_ylim([0, 1.0])
fig.set_size_inches(10, 5, forward=True)
ax[0].set_title("GNS")
ax[1].set_title("Ground Truth")

# Determine the rollout length
rollout_len = rollout["predicted_rollout"].shape[0] - 1

# Scatter plots for initial frame
scat0 = ax[0].scatter(
    rollout["predicted_rollout"][0, :, 0],
    rollout["predicted_rollout"][0, :, 1],
)
scat1 = ax[1].scatter(
    rollout["ground_truth_rollout"][0, :, 0],
    rollout["ground_truth_rollout"][0, :, 1],
)


# Animation function
def animate(i):
    scat0.set_offsets(rollout["predicted_rollout"][i])
    scat1.set_offsets(rollout["ground_truth_rollout"][i])
    return scat0, scat1


# Create the animation
ani = animation.FuncAnimation(fig, animate, repeat=True, frames=rollout_len, interval=50)

# Close the figure
plt.close(fig)

# Save the animation as a GIF
output_dir = "media"
output_path = os.path.join(output_dir, "scatter_tgv2d.gif")
os.makedirs(output_dir, exist_ok=True)  # Ensure the output directory exists
writer = animation.PillowWriter(fps=10, metadata=dict(artist="Me"), bitrate=1800)
ani.save(output_path, writer=writer)

# Display the GIF in the notebook
Image(url=output_path)
```
